## 2. Load the prepared example data

The dataset contains cropped donor and acceptor images, cell masks, and precomputed FRET measurements. The paths below are relative to this notebook.

To adjust video frame rate, marker color, image intensity ranges, or the FRET pseudocolor scale, edit the display settings in `functions.py`.

````python
from pathlib import Path
from functions import (
    FRETVisualization,
    VisualizationSettings,
    show_interactive,
    export_video,
)

DATA_DIR = Path("example_data/mouse6")
OUTPUT_DIR = Path("outputs")

analysis = FRETVisualization(
    DATA_DIR, sample_name="mouse6", settings=VisualizationSettings()
)
print(f"Input frames: {len(analysis.available_T)}")
print(f"FRET baseline: {analysis.fret_baseline:.6f}")
````

## 3. Preview the visualization

The upper panels display the donor, acceptor, and FRET pseudocolor images. The lower panel shows the mean FRET change with a cell-level error band and a marker at the displayed time point.

````python
import matplotlib.pyplot as plt

fig = analysis.plot_combined_frame(0)
plt.show()
````

## 4. Export the video visualization and mean FRET data

The movie uses the prepared visualization frames in sequence, skipping unavailable raw acquisition frames. Only the finished MP4 and one CSV are saved in `outputs/`.

- `pHLr_analysis_video_results.mp4` — final visualization
- `mean_FRET.csv` — `time_sec`, `mean_FRET`, `error_FRET`

The error represents variability among measured cells within this mouse, not across mice. Leave `VIDEO_FRAME_INDICES = None` for the complete movie.


````python
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

VIDEO_FRAME_INDICES = None  # Full movie

video_path = export_video(
    analysis, OUTPUT_DIR / "pHLr_analysis_video_results.mp4",
    indices=VIDEO_FRAME_INDICES,
)
csv_path = analysis.save_curve_csv(OUTPUT_DIR / "mean_FRET.csv")

print("Saved:", video_path)
print("Saved:", csv_path)
````

## 5. Play the saved video

This player opens the **exported MP4** above, not the original acquisition frames or the interactive frame-by-frame preview. Missing raw acquisition frames are not shown as blank intervals; playback follows the frames included in the finished video.


````python
from IPython.display import Video, display

display(Video(filename=str(video_path), embed=True, html_attributes="controls"))
````
